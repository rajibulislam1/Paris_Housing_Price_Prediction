```python
def column_summary_light(df):
    return pd.DataFrame({
        "missing_values": df.isna().sum(),
        "num_unique": df.nunique(),
        "dtype": df.dtypes
    }).reset_index().rename(columns={"index": "column"})
```

```python
column_summary_light(df)
```
